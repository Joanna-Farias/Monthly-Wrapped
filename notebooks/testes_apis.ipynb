```python
import os
import json
import requests
from dotenv import load_dotenv

load_dotenv()
LASTFM_KEY = os.getenv("LASTFM_API_KEY")
LASTFM_USER = os.getenv("LASTFM_USERNAME")
print(bool(LASTFM_KEY), LASTFM_USER)  # confere sem expor a chave
```

```python
def lastfm(method, **params):
    p = {"method": method, "user": LASTFM_USER, "api_key": LASTFM_KEY,
         "format": "json", **params}
    r = requests.get("https://ws.audioscrobbler.com/2.0/", params=p, timeout=15)
    r.raise_for_status()
    return r.json()

top_tracks = lastfm("user.gettoptracks", period="1month", limit=30)
top_artists = lastfm("user.gettopartists", period="1month", limit=30)
```

```python
os.makedirs("../data/raw", exist_ok=True)
with open("../data/raw/lastfm_top_tracks.json", "w", encoding="utf-8") as f:
    json.dump(top_tracks, f, ensure_ascii=False, indent=2)

tracks = top_tracks["toptracks"]["track"]
print(json.dumps(tracks[0], indent=2, ensure_ascii=False))
```
